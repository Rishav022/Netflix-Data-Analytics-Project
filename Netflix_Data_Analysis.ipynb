```python
df_clean["month_added"] = df_clean["date_added"].dt.month_name()
```

```python
## count the content produced per month

month_count = df_clean["month_added"].value_counts()
```

```python
## value count return the value in descending order but we want months in their chronological order , so arrange the

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
## reindex will arrange them acc to month-order
month_count = month_count.reindex(month_order)
month_count
```

```python
## plot a line chart for this

month_count.plot(
    kind="line",
    marker="o",
    markersize=6,
    markerfacecolor="orange",
    markeredgecolor="orange",
    figsize=(10, 5)
)

plt.grid(True)
plt.title("Monthly Content Additions on Netflix")
plt.xlabel("Month")
plt.ylabel("Number of Titles")
plt.xticks(ticks=range(len(month_count)), labels=month_count.index, rotation=45)
plt.savefig("images/monthly_content.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## find the duration of content

##1.splitting movies from tv shows because movies are in minutes

movies = df_clean[df_clean["type"] == "Movie"].copy()

## remove the min from the duration and convert it to int dtype
movies["duration_min"] = (
    movies["duration"]
    .str.replace(" min", "", regex=False)
    .astype(int)
)
```

```python
## plotting a histogram for this data

plt.figure(figsize=(10, 6))

movies["duration_min"].hist(
    bins=20,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Movie Durations on Netflix")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")
plt.grid(True, linestyle="--", alpha=0.5)
plt.savefig("images/duration_ditribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## find the genres produces per year

genre_year = df_clean[["listed_in", "year_added"]].copy()

# Split the genres
genre_year["listed_in"] = genre_year["listed_in"].str.split(", ")

# Explode the genres
genre_year = genre_year.explode("listed_in")

# Create the cross-tab
genre_year_table = pd.crosstab(
    genre_year["year_added"],
    genre_year["listed_in"]
)

# Find the top 5 genres
top5_genres = genre_year["listed_in"].value_counts().head(5)

# Keep only the top 5 genre columns
top5_genre_trend = genre_year_table.loc[:, top5_genres.index]
```

```python
## plot the line graph for this

top5_genre_trend.plot(
    kind="line",
    figsize=(12, 6),
    marker="o"
)

plt.title("Trend of Top 5 Netflix Genres Over the Years")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")

plt.xticks(top5_genre_trend.index, rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend(title="Genre")
plt.savefig("images/genre_trends.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## country produce average longest movie

movie_country = movies[["country","duration_min"]].copy()
movie_country = movie_country[movie_country["country"]!="Unknown"]
movie_country["country"] = movie_country["country"].str.split(", ")
movie_country = movie_country.explode("country")
avg_duration = (
    movie_country
    .groupby("country")["duration_min"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
avg_duration.plot(
    kind="barh",
    figsize=(10, 6),
    color="skyblue",
    edgecolor="black"
)

plt.title("Top 10 Countries with the Longest Average Movie Duration")
plt.xlabel("Average Duration (Minutes)")
plt.ylabel("Country")
plt.grid(True, axis="x", linestyle="--", alpha=0.5)
plt.gca().invert_yaxis()
plt.savefig("images/top_average_duration.png", dpi=300, bbox_inches="tight")
plt.show()
```
